### 1. Library Import and Image Loading
* **`import cv2`**: **Imports the OpenCV library, which is the primary tool used for computer vision tasks like reading and transforming images.**
* **`cv2.imread()`**: **Reads the image file from your storage and converts it into a numerical matrix or array.**
* **Channel Order**: **Note that OpenCV loads images in BGR (Blue-Green-Red) format by default, which is different from the standard RGB order.**

### 2. Numerical Representation and Shape
* **`img`**: **Displays the raw pixel intensity values. In a computer, every image is simply a grid of numbers representing colors.**
* **`img.shape`**: **Provides the dimensions of the image in the format (Height, Width, Channels). The '3' indicates a color image.**



### 3. Color Space Conversion
* **`import matplotlib.pyplot as plt`**: **Imports the Matplotlib library used to visualize data and display images within the notebook.**
* **`cv2.cvtColor(img, cv2.COLOR_BGR2RGB)`**: **Converts the image from BGR to RGB format so that Matplotlib can display the colors correctly.**



### 4. Image Resizing
* **`cv2.resize(img, (16, 16))`**: **Standardizes the image to a fixed size of 16x16 pixels. This is a required step because Neural Networks need a consistent input size.**

```python
# IMAGE PREPROCESSING
## Resizing an image
import cv2
import matplotlib.pyplot as plt

# Load an image
img = cv2.imread("/content/tree-clipart-xl.png")   # Changed from "/content/cat.jpeg" to an existing image
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Convert BGR → RGB

# Resize to 64x64
resized = cv2.resize(img, (240, 360))

plt.subplot(1,2,1)
plt.title("Original")
plt.imshow(img)

plt.subplot(1,2,2)
plt.title("Resized 240x240")
plt.imshow(resized)
plt.show()
```

```python
img.shape
```

```python
rotated = cv2.rotate(img, cv2.ROTATE_90_COUNTERCLOCKWISE)

# Show images
plt.subplot(1,2,1); plt.imshow(img); plt.title("Original"); plt.axis("off")
plt.subplot(1,2,2); plt.imshow(rotated); plt.title("Rotated 270°"); plt.axis("off")
plt.show()
```

```python
### HORIZONTAL FLIP

flipped = cv2.flip(img,1)

# Show images
plt.subplot(1,2,1); plt.imshow(img); plt.title("Original"); plt.axis("off")
plt.subplot(1,2,2); plt.imshow(flipped); plt.title("Horizontal Flip"); plt.axis("off")
plt.show()
```

```python
# CROPING AN IMAGE
# Crop (top-left corner: (50,50), bottom-right corner: (200,200))
cropped = img[1350:1500, 1200:1400]

# Show original and cropped image
plt.subplot(1,2,1); plt.imshow(img); plt.title("Original"); plt.axis("off")
plt.subplot(1,2,2); plt.imshow(cropped); plt.title("Cropped"); plt.axis("off")
plt.show()
```
